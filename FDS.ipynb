```python
# Experiment 13: Clean missing and inappropriate values in Hotel.csv
df = pd.read_csv("Hotel.csv")
print("Duplicates before cleaning:", int(df.duplicated().sum()))
df = df.drop_duplicates().copy().reset_index(drop=True)
# Remove the redundant duplicated age column if present.
df = df.loc[:, ~df.columns.duplicated()]
if "Age_Group.1" in df.columns:
    df = df.drop(columns="Age_Group.1")

for column in ["CustomerID", "Rating(1-5)", "Bill", "NoOfPax", "EstimatedSalary"]:
    df[column] = pd.to_numeric(df[column], errors="coerce")
df.loc[df["CustomerID"] < 0, "CustomerID"] = np.nan
df.loc[~df["Rating(1-5)"].between(1, 5), "Rating(1-5)"] = np.nan
df.loc[df["Bill"] < 0, "Bill"] = np.nan
df.loc[df["EstimatedSalary"] < 0, "EstimatedSalary"] = np.nan
df.loc[~df["NoOfPax"].between(1, 20), "NoOfPax"] = np.nan
df["Hotel"] = df["Hotel"].replace({"Ibys": "Ibis"})
df["FoodPreference"] = df["FoodPreference"].replace(
    {"Vegetarian": "Veg", "veg": "Veg", "non-Veg": "Non-Veg"}
)
for column in ["CustomerID", "Bill", "EstimatedSalary"]:
    df[column] = df[column].fillna(df[column].mean())
for column in ["NoOfPax", "Rating(1-5)"]:
    df[column] = df[column].fillna(df[column].median())
for column in ["Age_Group", "Hotel", "FoodPreference"]:
    if df[column].isna().any():
        df[column] = df[column].fillna(df[column].mode().iloc[0])
print("Cleaned data:\n", df)
print("Remaining missing values:\n", df.isna().sum())
```

```python
np.random.seed(1)

# Step 1: Population of delivery times (minutes)
population = np.random.normal(loc=40, scale=8, size=100000)
mu, sigma = population.mean(), population.std()
print(f"Population mean = {mu:.2f} min, Population SD = {sigma:.2f} min\n")

# Step 2: Take 1000 samples for each sample size and store their means
sample_sizes = [10, 30, 100]
num_samples = 1000
fig, ax = plt.subplots(2, 2, figsize=(12, 8))
ax = ax.ravel()

ax[0].hist(population, bins=50, color='lightgray', edgecolor='black')
ax[0].axvline(mu, color='red', linestyle='--', label=f'Population mean = {mu:.1f}')
ax[0].set_title('Population: Delivery Times of 1,00,000 Parcels')
ax[0].set_xlabel('Minutes')
ax[0].set_ylabel('Frequency')
ax[0].legend()

print(f"{'n':>5} {'Mean of sample means':>22} {'SE = σ/√n':>12} {'SD of sample means':>20}")
for i, n in enumerate(sample_sizes):
    sample_means = [np.random.choice(population, size=n, replace=False).mean() for _ in range(num_samples)]
    se_formula = sigma / np.sqrt(n)
    se_observed = np.std(sample_means)
    print(f"{n:>5} {np.mean(sample_means):>22.2f} {se_formula:>12.2f} {se_observed:>20.2f}")
    ax[i + 1].hist(sample_means, bins=30, color='lightgreen', edgecolor='black')
    ax[i + 1].axvline(mu, color='red', linestyle='--', label='Population mean')
    ax[i + 1].set_xlim(30, 50)  # same scale so the spread can be compared
    ax[i + 1].set_title(f'Sampling Distribution of Mean (n = {n}), SE = {se_formula:.2f}')
    ax[i + 1].set_xlabel('Sample mean (minutes)')
    ax[i + 1].set_ylabel('Frequency')
    ax[i + 1].legend()

plt.tight_layout()
plt.show()

# Step 3: SE estimated from ONE sample of 30 parcels (sigma not known)
one_sample = np.random.choice(population, size=30, replace=False)
se_est = one_sample.std(ddof=1) / np.sqrt(30)
print(f"\nOne sample of 30 parcels: mean = {one_sample.mean():.2f}, estimated SE = s/√n = {se_est:.2f}")
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv("Salary_data.csv")
print(df.info())

df = df[["YearsExperience", "Salary"]].apply(
    pd.to_numeric, errors="coerce"
).dropna()

X = df["YearsExperience"].to_numpy()
y = df["Salary"].to_numpy()

# Deterministic 80/20 train-test split
indices = np.random.default_rng(0).permutation(len(y))
test_count = max(1, round(0.2 * len(y)))
test_indices = indices[:test_count]
train_indices = indices[test_count:]

x_train, x_test = X[train_indices], X[test_indices]
y_train, y_test = y[train_indices], y[test_indices]

# Fit intercept and slope using least squares
train_matrix = np.column_stack((np.ones(len(x_train)), x_train))
intercept, coefficient = np.linalg.lstsq(train_matrix, y_train, rcond=None)[0]

def predict(x):
    return intercept + coefficient * x

def r2_score(actual, predicted):
    return 1 - np.sum((actual - predicted) ** 2) / np.sum((actual - actual.mean()) ** 2)

print(f"Training R² score: {r2_score(y_train, predict(x_train)):.4f}")
print(f"Testing R² score: {r2_score(y_test, predict(x_test)):.4f}")
print(f"Coefficient: {coefficient:.2f}")
print(f"Intercept: {intercept:.2f}")
```

```python
# Experiment 6: Website traffic time-series analysis
rng = np.random.default_rng(6)
dates = pd.date_range("2025-01-01", periods=180, freq="D")
day_effect = np.array([0, 1, 2, 3, 4, -8, -12])[dates.dayofweek]
page_views = 800 + np.arange(len(dates)) * 1.3 + 55 * np.sin(2 * np.pi * np.arange(len(dates)) / 30) + day_effect * 5 + rng.normal(0, 24, len(dates))
page_views[100] += 240  # deliberate spike for anomaly detection
website = pd.DataFrame({
    "Page_Views": page_views,
    "Unique_Visitors": page_views * rng.uniform(0.62, 0.77, len(dates)),
    "Bounce_Rate": np.clip(0.48 + 0.04 * np.sin(2 * np.pi * np.arange(len(dates)) / 7) + rng.normal(0, 0.025, len(dates)), 0.2, 0.8),
}, index=dates)
website.index.name = "Date"
website.iloc[[12, 49], 0] = np.nan
website["Page_Views"] = website["Page_Views"].interpolate(method="time")
website["Unique_Visitors"] = website["Unique_Visitors"].interpolate(method="time")
website["Bounce_Rate"] = website["Bounce_Rate"].interpolate(method="time")
print(website.head())

# A centered 7-day moving average estimates trend; weekday averages estimate seasonality.
website["Trend"] = website["Page_Views"].rolling(window=7, center=True, min_periods=1).mean()
weekday_trend = website["Page_Views"] - website["Trend"]
seasonal_pattern = weekday_trend.groupby(website.index.dayofweek).mean()
website["Seasonality"] = website.index.dayofweek.map(seasonal_pattern)
website["Residual"] = website["Page_Views"] - website["Trend"] - website["Seasonality"]
website["Page_Views_MA30"] = website["Page_Views"].rolling(window=30, min_periods=1).mean()
website["Page_Views_Z"] = (website["Page_Views"] - website["Page_Views"].mean()) / website["Page_Views"].std()
anomalies = website[website["Page_Views_Z"].abs() > 3]
print("Anomaly dates:\n", anomalies[["Page_Views", "Page_Views_Z"]])

fig, axes = plt.subplots(3, 1, figsize=(12, 10), sharex=True)
website[["Page_Views", "Unique_Visitors"]].plot(ax=axes[0], title="Website Traffic")
website[["Page_Views", "Page_Views_MA30"]].plot(ax=axes[1], title="Page Views and 30-day Moving Average")
website[["Trend", "Seasonality", "Residual"]].plot(ax=axes[2], title="Trend, Weekday Seasonality, and Residual")
axes[0].scatter(anomalies.index, anomalies["Page_Views"], color="red", label="Z-score anomaly")
axes[0].legend()
plt.tight_layout()
plt.show()
print("Mean bounce rate:", round(website["Bounce_Rate"].mean(), 3))
```

```python
# Experiment 8: Two-tailed z-test for bird weights
import math

sample_data = np.array([152, 148, 151, 149, 147, 153, 150, 148, 152, 149, 151, 150, 149, 152, 151, 148, 150, 152, 149, 150, 148, 153, 151, 150, 149, 152, 148, 151, 150, 153], dtype=float)
null_mean = 150.0
sample_mean = sample_data.mean()
sample_std = sample_data.std(ddof=1)
n = sample_data.size
z_statistic = (sample_mean - null_mean) / (sample_std / np.sqrt(n))
normal_cdf = 0.5 * (1 + math.erf(abs(z_statistic) / np.sqrt(2)))
p_value = 2 * (1 - normal_cdf)
alpha = 0.05
print(f"Sample mean: {sample_mean:.2f} g")
print(f"Z-statistic: {z_statistic:.4f}")
print(f"Two-tailed p-value: {p_value:.4f}")
if p_value < alpha:
    print("Reject H0: average weight differs significantly from 150 g.")
else:
    print("Fail to reject H0: insufficient evidence that average weight differs from 150 g.")
```

```python
# Experiment 12: Detect and remove outliers using the IQR rule
rng = np.random.default_rng(12)
measurements = np.concatenate([rng.integers(20, 81, size=30), [180]])
q1, q3 = np.percentile(measurements, [25, 75])
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outlier_mask = (measurements < lower_bound) | (measurements > upper_bound)
print(f"Q1={q1:.2f}, Q3={q3:.2f}, IQR={iqr:.2f}")
print(f"Outlier bounds: [{lower_bound:.2f}, {upper_bound:.2f}]")
print("Detected outliers:", measurements[outlier_mask])
filtered_measurements = measurements[~outlier_mask]
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
sns.boxplot(x=measurements, ax=axes[0])
axes[0].set_title("Before Outlier Removal")
sns.boxplot(x=filtered_measurements, ax=axes[1])
axes[1].set_title("After Outlier Removal")
plt.tight_layout()
plt.show()
```

```python
# Experiment 18: K-nearest neighbors classification with the provided Iris CSV
iris_df = pd.read_csv("sample1.csv")
feature_columns = ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
X = iris_df[feature_columns].to_numpy(dtype=float)
y = iris_df["variety"].to_numpy()
rng = np.random.default_rng(18)
indices = rng.permutation(len(y))
test_size = int(round(0.2 * len(y)))
test_indices, train_indices = indices[:test_size], indices[test_size:]
X_train, X_test = X[train_indices], X[test_indices]
y_train, y_test = y[train_indices], y[test_indices]
mean = X_train.mean(axis=0)
scale = X_train.std(axis=0)
scale[scale == 0] = 1
X_train_scaled = (X_train - mean) / scale
X_test_scaled = (X_test - mean) / scale

def knn_predict(x_train, y_train, x_values, k=5):
    predictions = []
    for row in x_values:
        nearest = np.argsort(np.sum((x_train - row) ** 2, axis=1))[:k]
        labels, counts = np.unique(y_train[nearest], return_counts=True)
        predictions.append(labels[np.argmax(counts)])
    return np.asarray(predictions)

y_pred = knn_predict(X_train_scaled, y_train, X_test_scaled, k=5)
accuracy = np.mean(y_pred == y_test)
classes = np.unique(y)
confusion = pd.DataFrame(0, index=classes, columns=classes)
for actual, predicted in zip(y_test, y_pred):
    confusion.loc[actual, predicted] += 1
print(f"Test accuracy: {accuracy:.3f}")
print("Confusion matrix (rows=actual, columns=predicted):\n", confusion)
```

```python
# Experiment 19: Logistic regression classification (synthetic local demonstration)
rng = np.random.default_rng(19)
row_count = 400
age = rng.integers(18, 61, row_count).astype(float)
salary = rng.integers(15000, 150001, row_count).astype(float)
logit = -7.0 + 0.14 * (age - 35) + 0.000035 * (salary - 70000)
probability = 1 / (1 + np.exp(-logit))
purchased = rng.binomial(1, probability)
X = np.column_stack([age, salary])
mean, scale = X.mean(axis=0), X.std(axis=0)
X = (X - mean) / scale
X = np.column_stack([np.ones(row_count), X])
rng = np.random.default_rng(190)
indices = rng.permutation(row_count)
test_indices = indices[:80]
train_indices = indices[80:]
X_train, X_test = X[train_indices], X[test_indices]
y_train, y_test = purchased[train_indices], purchased[test_indices]
weights = np.zeros(X_train.shape[1])
learning_rate = 0.1
for _ in range(3000):
    scores = np.clip(X_train @ weights, -30, 30)
    predicted_probability = 1 / (1 + np.exp(-scores))
    gradient = X_train.T @ (predicted_probability - y_train) / len(y_train)
    weights -= learning_rate * gradient
y_pred = (1 / (1 + np.exp(-np.clip(X_test @ weights, -30, 30))) >= 0.5).astype(int)
print(f"Test accuracy: {np.mean(y_pred == y_test):.3f}")
print("Confusion matrix (rows=actual, columns=predicted):\n", pd.crosstab(pd.Series(y_test, name="Actual"), pd.Series(y_pred, name="Predicted")))
print("Learned standardized-feature coefficients:", weights.round(3))
```

```python
# Experiment 20: K-means clustering of customer income and spending (synthetic local demo)
rng = np.random.default_rng(20)
cluster_centers = np.array([[25, 25], [25, 75], [50, 50], [75, 25], [75, 75]], dtype=float)
customer_features = np.vstack([center + rng.normal(0, 7, size=(40, 2)) for center in cluster_centers])
customer_features = np.clip(customer_features, 1, 100)
cluster_count = 5
centers = customer_features[rng.choice(len(customer_features), cluster_count, replace=False)].copy()
for _ in range(100):
    distances = np.linalg.norm(customer_features[:, None, :] - centers[None, :, :], axis=2)
    labels = distances.argmin(axis=1)
    updated_centers = np.array([customer_features[labels == group].mean(axis=0) if np.any(labels == group) else centers[group] for group in range(cluster_count)])
    if np.allclose(updated_centers, centers):
        break
    centers = updated_centers
inertia = np.sum((customer_features - centers[labels]) ** 2)
print("Cluster sizes:", np.bincount(labels, minlength=cluster_count))
print("Cluster centers (income-like, spending score):\n", centers.round(2))
print(f"Within-cluster sum of squares (inertia): {inertia:.2f}")
plt.figure(figsize=(8, 6))
plt.scatter(customer_features[:, 0], customer_features[:, 1], c=labels, cmap="tab10", alpha=0.75)
plt.scatter(centers[:, 0], centers[:, 1], marker="X", s=220, c="black", label="Centroids")
plt.xlabel("Annual Income (scaled units)")
plt.ylabel("Spending Score")
plt.title("K-means Customer Clusters")
plt.legend()
plt.show()
```
